###  Estimate pi

3. Obtain an estimate of the value of $\pi$
    - From the definition in `numpy`
    - Using an analytic series expansion: $\pi = 4 \sum_{k=0}^\infty \frac{(-1)^{k}}{2k + 1}$
    - Using a Monte-Carlo estimate: 
        1. Sample $n$ points uniformly on the unit square using `numpy.random.uniform`
        2. Count the number $n_{<r}$ of points with distance $r = \sqrt{x^2+y^2} \leq 1$
        3. Compare the fraction $\frac{n_{<r}}{n}$ to the ratio of the area of a quadrant of a disc to the area of the square to derive an estimate of $\pi$
    - Bonus: Repeat for (hyper) cube and (hyper) ball in 3 or more dimensions. How does the number of samples required to get a robust estimate of $\pi$ change with the number of dimensions?

Use Leibniz and Nilakantha

```python
up_lim = 5000
pi_an = 4 * sum((-1)**k/(2*k+1) for k in range (0,up_lim +1))
pi_extra = 3 + sum(4 * (-1)**(n+1) / ((2*n)*(2*n+1)*(2*n+2)) for n in range(1, up_lim+1))
print(pi_an)
print(pi_extra)
```

```python
def pi_an(up_lim):
    return 4 * sum((-1)**k/(2*k+1) for k in range (0,up_lim +1)) 
def pi_extra(up_lim):
    return 3 + sum(4 * (-1)**(n+1) / ((2*n)*(2*n+1)*(2*n+2)) for n in range(1, up_lim+1))
```

```python
import matplotlib.pyplot as plt
```

```python
up_lim = range(1, 101)

y1 = [pi_an(k) for k in up_lim]
y2 = [pi_extra(k) for k in up_lim]

plt.plot(up_lim, y1, label="Leibniz")
plt.plot(up_lim, y2, label="Nilakantha")
plt.axhline(np.pi, color='r', linestyle='--', label='π')
plt.title("Analytic pi estimate")

plt.legend()
plt.show()
```

```python
up_lim = range(1, 101)

y1 = [pi_an(k)-np.pi for k in up_lim]
y2 = [pi_extra(k)-np.pi for k in up_lim]

plt.plot(up_lim, y1, label="Leibniz")
plt.plot(up_lim, y2, label="Nilakantha")
plt.title("residuals")

plt.legend()
plt.show()
```

### Monte Carlo

```python
import numpy as np

n = 100000

# 1. Generate n random points in the unit square
x = np.random.uniform(-1, 1, n)
y = np.random.uniform(-1, 1, n)

# 2. Count points inside the quarter circle
inside = x**2 + y**2 <= 1
n_inside = np.sum(inside)

# 3. Estimate pi
pi_estimate = 4 * n_inside / n

print("Estimated pi:", pi_estimate)
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = 10000

x = np.random.uniform(-1, 1, n)
y = np.random.uniform(-1, 1, n)

inside = x**2 + y**2 <= 1

plt.scatter(x[inside], y[inside], s=5, label="Inside")
plt.scatter(x[~inside], y[~inside], s=5, label="Outside")

plt.xlabel("x")
plt.ylabel("y")
plt.axis("equal")
plt.legend()
plt.show()
```
