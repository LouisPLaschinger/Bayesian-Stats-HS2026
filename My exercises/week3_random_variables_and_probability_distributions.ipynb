### Exercise 1

Consider the distribution with PDF
$$
    p(x) = \begin{cases}
                \frac{1}{2}\sin x & 0 \le x \le \pi \\
                0 & \text{otherwise}
            \end{cases}
$$

1. Derive the CDF $F(x) = \int_{-\infty}^{x} p(x')\,\dd x'$ analytically, and plot it together with the PDF $p(x)$.
2. Find the inverse CDF $F^{-1}(u)$, and use it to transform samples $U \sim \Uniform(0, 1)$ into samples $X = F^{-1}(U)$ (inverse transform sampling).
3. Compare a density-normalised histogram of the samples with the PDF $p(x)$.

$$F(x) = \int_{-\infty}^{x} p(x')\,d x' = \int_{-\infty}^{x} \frac{1}{2}sin(x')|_0^\pi\,d x' = \frac{1}{2}(1-cos(x)|_0^\pi)$$
0 and 1 outside of that.

```python
import numpy as np
import matplotlib.pyplot as plt

# x-Werte
x = np.linspace(-1, np.pi + 1, 1000)

# Dichte p(x)
p = np.where((x >= 0) & (x <= np.pi), 0.5 * np.sin(x), 0)

# Verteilungsfunktion F(x)
F = np.where(x < 0,0, np.where(x <= np.pi,0.5 * (1 - np.cos(x)), 1))

# Plot
plt.figure(figsize=(10, 5))

plt.plot(x, p, label=r"$p(x)$")
plt.plot(x, F, label=r"$F(x)$")

plt.xlabel(r"$x$")
plt.ylabel("Value")
plt.title("p(x) F(x)")

plt.legend()

plt.grid()
plt.show()
```

$$F^{-1}(U)= cos^{-1}(1-2U)$$

```python
u = np.random.uniform(0, 1, 7000)
x = np.arccos(1 - 2*u)

plt.hist(x, bins=30, density=True)
plt.xlabel("x")
plt.ylabel("density")

x_plot = np.linspace(0, np.pi, 1000)
p = 0.5 * np.sin(x_plot)

plt.plot(x_plot, p, ls='--',label=r"$p(x)=\frac{1}{2}\sin(x)$")

plt.legend()

plt.show()
```

1. Confirm by simulation that, for large $n$ and small $\Delta x$ with the rate $\lambda = n\Delta x$ held fixed, $K$ follows a Poisson distribution:

$$
\Pr(K=k)=\frac{\lambda^k}{k!}e^{-\lambda}.
$$

2. Derive the Poisson distribution from the binomial distribution,

$$
\Pr(K=k)=\binom{n}{k}p^k(1-p)^{n-k},
$$

in the limit $n\to\infty$ with $np=\lambda$ held fixed.
